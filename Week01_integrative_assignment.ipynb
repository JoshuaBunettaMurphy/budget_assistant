## Part 2: Revenue Analysis (25 points)

David pulls up NovaTech's subscription data. *"This is the heart of our business model,"* he explains. *"We have three pricing tiers, and investors will want to understand our revenue composition."*

**Subscription Tiers:**

| Tier | Monthly Price | Number of Customers |
|------|--------------|--------------------|
| Starter | $49.00 | 312 |
| Professional | $149.00 | 156 |
| Enterprise | $499.00 | 38 |

**Your Task:**
1. Calculate Monthly Recurring Revenue (MRR) for each tier
2. Calculate Total MRR
3. Calculate Annual Recurring Revenue (ARR) = MRR × 12
4. Calculate the percentage of total MRR that comes from each tier
5. Calculate the total number of customers
6. Calculate Average Revenue Per User (ARPU) = Total MRR / Total Customers

**Display a formatted revenue report that Marcus can use in the pitch.**

```python
# Part 2: Revenue Analysis
# Calculate NovaTech's key revenue metrics

# Subscription tier data
starter_price = 49.00
starter_customers = 312

professional_price = 149.00
professional_customers = 156

enterprise_price = 499.00
enterprise_customers = 38

# Calculate MRR per tier
starter_mrr = starter_price * starter_customers
professional_mrr = professional_price * professional_customers
enterprise_mrr = enterprise_price * enterprise_customers

# Calculate total MRR and ARR
total_mrr = starter_mrr + professional_mrr + enterprise_mrr
annual_recurring_revenue = total_mrr * 12

# Calculate percentages
starter_mrr_percent = (starter_mrr / total_mrr) * 100
professional_mrr_percent = (professional_mrr / total_mrr) * 100
enterprise_mrr_percent = (enterprise_mrr / total_mrr) * 100

# Calculate customer metrics
total_customers = starter_customers + professional_customers + enterprise_customers
arpu = total_mrr / total_customers

# Print formatted revenue report
print("="*60)
print("              REVENUE ANALYSIS")
print("="*60)
print(f"Monthly Recurring Revenue (MRR):")
print(f"  Starter Tier:       ${starter_mrr:,.2f} ({starter_mrr_percent:.2f}%)")
print(f"  Professional Tier:  ${professional_mrr:,.2f} ({professional_mrr_percent:.2f}%)")
print(f"  Enterprise Tier:    ${enterprise_mrr:,.2f} ({enterprise_mrr_percent:.2f}%)")
print("-" * 60)
print(f"Total MRR:            ${total_mrr:,.2f}")
print(f"Annual Recurring Revenue (ARR): ${annual_recurring_revenue:,.2f}")
print("-" * 60)
print(f"Total Customers:      {total_customers:,}")
print(f"Average Revenue Per User (ARPU): ${arpu:,.2f}")
print("="*60)
```

## Part 3: Growth Metrics (25 points)

David slides another sheet across the desk. *"Investors love growth stories. Here's our trajectory over the past few months."*

**Monthly Performance Data:**

| Month | New Customers | Churned Customers | MRR at Month End |
|-------|--------------|-------------------|------------------|
| October 2025 | 28 | 5 | $68,420.00 |
| November 2025 | 35 | 7 | $71,850.00 |
| December 2025 | 42 | 4 | $76,290.00 |
| January 2026 | 51 | 6 | $81,634.00 |

**Your Task:**
1. Calculate Net New Customers for each month (New - Churned)
2. Calculate the MRR growth rate from October to January: ((Jan MRR - Oct MRR) / Oct MRR) × 100
3. Calculate total customers acquired over the 4 months
4. Calculate total customers lost (churned) over the 4 months
5. Calculate the average monthly churn rate: (Total Churned / Total Acquired) × 100
6. Determine if NovaTech is "growing fast" (net new customers per month > 20 on average)

**Create a growth summary for the investor presentation.**

```python
# Part 3: Growth Metrics
# Analyze NovaTech's growth trajectory

# Monthly data
oct_new_customers = 28
oct_churned_customers = 5
oct_mrr = 68420.00

nov_new_customers = 35
nov_churned_customers = 7
nov_mrr = 71850.00

dec_new_customers = 42
dec_churned_customers = 4
dec_mrr = 76290.00

jan_new_customers = 51
jan_churned_customers = 6
jan_mrr = 81634.00

# Calculate net new customers per month
oct_net_new_customers = oct_new_customers - oct_churned_customers
nov_net_new_customers = nov_new_customers - nov_churned_customers
dec_net_new_customers = dec_new_customers - dec_churned_customers
jan_net_new_customers = jan_new_customers - jan_churned_customers

# Calculate MRR growth rate
mrr_growth_rate = ((jan_mrr - oct_mrr) / oct_mrr) * 100

# Calculate acquisition and churn totals
total_acquired_customers = oct_new_customers + nov_new_customers + dec_new_customers + jan_new_customers
total_churned_customers = oct_churned_customers + nov_churned_customers + dec_churned_customers + jan_churned_customers

# Calculate average monthly churn rate
# Assuming total acquired customers is the base for churn rate calculation over the period
# If total customers at start of period is needed, it's not provided in growth metrics
average_monthly_churn_rate = (total_churned_customers / total_acquired_customers) * 100

# Determine if 'growing fast'
average_net_new_customers_per_month = (oct_net_new_customers + nov_net_new_customers + dec_net_new_customers + jan_net_new_customers) / 4
is_growing_fast = average_net_new_customers_per_month > 20

# Print growth summary
print("="*60)
print("              GROWTH METRICS")
print("           October 2025 - January 2026")
print("="*60)
print(f"Net New Customers:")
print(f"  October 2025:       {oct_net_new_customers}")
print(f"  November 2025:      {nov_net_new_customers}")
print(f"  December 2025:      {dec_net_new_customers}")
print(f"  January 2026:       {jan_net_new_customers}")
print("-" * 60)
print(f"MRR Growth Rate (Oct 2025 to Jan 2026): {mrr_growth_rate:.2f}%")
print("-" * 60)
print(f"Total Customers Acquired (4 months):  {total_acquired_customers}")
print(f"Total Customers Lost (4 months):    {total_churned_customers}")
print(f"Average Monthly Churn Rate:         {average_monthly_churn_rate:.2f}%")
print("-" * 60)
print(f"Average Net New Customers/Month:    {average_net_new_customers_per_month:.2f}")
print(f"Is NovaTech 'Growing Fast' (>20 avg): {'Yes' if is_growing_fast else 'No'}")
print("="*60)
```

## Part 4: Investment Readiness Check (20 points)

*"Before tomorrow's meeting,"* David says, *"we need to verify that we meet Apex Ventures' investment criteria. They have specific thresholds they look for."*

**Apex Ventures Investment Criteria:**

A company is considered "investment ready" if it meets **ALL** of the following:
1. ARR of at least $750,000
2. At least 400 total customers
3. MRR growth rate of at least 15% over the evaluation period
4. Average churn rate below 20%

Additionally, a company qualifies for their **"High Growth"** bonus investment tier if:
- They meet ALL basic criteria, AND
- Either ARR exceeds $1,000,000 OR they have more than 500 customers

**Your Task:**
Using the metrics you calculated in Parts 2 and 3:
1. Check each criterion individually and store the result (True/False)
2. Determine if NovaTech is investment ready (meets ALL criteria)
3. Determine if NovaTech qualifies for the "High Growth" tier
4. Create a clear "Investment Readiness Scorecard" showing pass/fail for each criterion

```python
# Part 4: Investment Readiness Check
# Evaluate NovaTech against Apex Ventures criteria

# Apex Ventures Investment Criteria
MIN_ARR = 750000
MIN_CUSTOMERS = 400
MIN_MRR_GROWTH_RATE = 15
MAX_CHURN_RATE = 20

# High Growth Bonus Criteria
HIGH_GROWTH_ARR = 1000000
HIGH_GROWTH_CUSTOMERS = 500

# Check each criterion (use variables from previous parts)
# annual_recurring_revenue, total_customers, mrr_growth_rate, average_monthly_churn_rate

criteria_arr_met = annual_recurring_revenue >= MIN_ARR
criteria_customers_met = total_customers >= MIN_CUSTOMERS
criteria_mrr_growth_met = mrr_growth_rate >= MIN_MRR_GROWTH_RATE
criteria_churn_rate_met = average_monthly_churn_rate < MAX_CHURN_RATE

# Determine investment readiness
is_investment_ready = criteria_arr_met and criteria_customers_met and criteria_mrr_growth_met and criteria_churn_rate_met

# Determine High Growth qualification
is_high_growth = is_investment_ready and \
                 (annual_recurring_revenue > HIGH_GROWTH_ARR or total_customers > HIGH_GROWTH_CUSTOMERS)

# Print Investment Readiness Scorecard
print("="*60)
print("         INVESTMENT READINESS SCORECARD")
print("            Apex Ventures Criteria")
print("="*60)
print(f"1. ARR of at least ${MIN_ARR:,.2f}:               {'PASSED' if criteria_arr_met else 'FAILED'} (Actual: ${annual_recurring_revenue:,.2f})")
print(f"2. At least {MIN_CUSTOMERS} total customers:      {'PASSED' if criteria_customers_met else 'FAILED'} (Actual: {total_customers:,})")
print(f"3. MRR growth rate of at least {MIN_MRR_GROWTH_RATE}%: {'PASSED' if criteria_mrr_growth_met else 'FAILED'} (Actual: {mrr_growth_rate:.2f}%)")
print(f"4. Average churn rate below {MAX_CHURN_RATE}%:       {'PASSED' if criteria_churn_rate_met else 'FAILED'} (Actual: {average_monthly_churn_rate:.2f}%)")
print("-" * 60)
print(f"Overall Investment Ready: {'YES' if is_investment_ready else 'NO'}")
print("-" * 60)
print(f"Qualifies for 'High Growth' tier: {'YES' if is_high_growth else 'NO'}")
print("="*60)
```
